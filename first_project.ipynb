```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales.drop_duplicates(inplace=True)
```

```python
sales['Status'].isnull().sum()
sales['unnamed1'].isnull().sum()
```

```python
sales.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
sales.columns
```

```python
sales['Amount'].isna().sum()
```

```python
sales.dropna(inplace=True)
```

```python
sales.info()
```

```python
sales.head()
```

```python
sales['Gender']=sales['Gender'].str.replace('F','female').str.replace('M','male')
```

```python
sales.head()
```

```python
sales['Marital_Status']=sales['Marital_Status'].astype(str)
```

```python
sales['Marital_Status']=sales['Marital_Status'].str.replace('0','unmarried').str.replace('1','married')
```

```python
sales.head()
```

```python
sales['State'].unique()
```

```python
def label(a):
    for i in a.containers:
        a.bar_label(i)
```

```python
fig,axes=plt.subplots(nrows=1,ncols=2,figsize=(20,10))
c_g=sns.countplot(x='Gender',data=sales,hue='Gender',ax=axes[0])
label(c_g)
```

```python
group_gender=sales.groupby('Gender')['Amount'].sum()
group_gender=pd.DataFrame(group_gender)
group_gender.reset_index(inplace=True)
group_gender
plt.pie(group_gender['Amount'],labels=group_gender['Gender'].unique(),
explode=(0.0,0.1),shadow=True,colors=('pink','brown'),autopct='%0.0f%%')
```

#### female is spending more money on shooping

```python
plt.figure(figsize=(20,10))
age_group=sales.groupby('Age Group')['Amount'].sum()
age_group=pd.DataFrame(age_group)
age_group.reset_index(inplace=True)
ab=sns.barplot(x='Age Group',y='Amount',data=age_group,hue='Age Group')
label(ab)
```

#### the age group between 25-35 is spending more money

```python
plt.figure(figsize=(20,10))
state_group=sales.groupby('State')['Amount'].sum()
state_group=pd.DataFrame(state_group)
state_group.reset_index(inplace=True)
plt.xticks(rotation=45,fontsize=15,color='green')
plt.xlabel('State',fontsize=15)
plt.ylabel('Amount',fontsize=20)
b=sns.barplot(x='State',y='Amount',data=state_group,hue='State',palette='coolwarm')
for i in b.containers:
    b.bar_label(i)
```

### Utter_predesh is spending more money

```python
plt.figure(figsize=(20,10))
Occupation_group=sales.groupby('Occupation')['Amount'].sum()
Occupation_group=pd.DataFrame(Occupation_group)
Occupation_group.reset_index(inplace=True)
plt.xticks(rotation=45,fontsize=15,color='green')
plt.xlabel('Occupation',fontsize=15)
plt.ylabel('Amount',fontsize=20)
c=sns.barplot(x='Occupation',y='Amount',data=Occupation_group,hue='Occupation',palette='magma')
for i in c.containers:
    c.bar_label(i)
```

#### IT sector is more spending money

```python
Marital_Status_group=sales.groupby('Marital_Status')['Amount'].sum()
Marital_Status_group=pd.DataFrame(Marital_Status_group)
Marital_Status_group.reset_index(inplace=True)
Marital_Status_group
plt.pie(Marital_Status_group['Amount'],labels=Marital_Status_group['Marital_Status'].unique(),
explode=(0.0,0.1),shadow=True,colors=('red','yellow'),autopct='%0.0f%%')
```

#### unmarried is spending more money on shppoing

```python
plt.figure(figsize=(20,10))
Product_Category_group=sales.groupby('Product_Category')['Amount'].sum()
Product_Category_group=pd.DataFrame(Product_Category_group)
Product_Category_group.reset_index(inplace=True)
plt.xticks(rotation=45,fontsize=15,color='green')
plt.xlabel('Occupation',fontsize=15)
plt.ylabel('Amount',fontsize=20)
d=sns.barplot(x='Product_Category',y='Amount',data=Product_Category_group,hue='Product_Category',palette='magma')
for i in d.containers:
    d.bar_label(i)
```
